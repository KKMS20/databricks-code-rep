** Use Case-1: Data Types and Casting **

```python
employee_name = input("Enter an Employee Name")
base_salary = input("Enter Basic Salary")
hra_percent = input("Enter HRA Percentage")
bonus_amount = input ("Enter Bonus Amount")
print (type(employee_name)); print (type(base_salary)); print (type(hra_percent)); print (type(bonus_amount));
hra = float(base_salary) * (int(hra_percent)/100)
total_salary = float(base_salary) + hra + float(bonus_amount)
print (float(base_salary)); print(hra); print (float(bonus_amount)); print (total_salary);
```

** Use Case-2: Data Types and Casting: Student Result Classification **

```python
marks = input("Enter the Marks")
print (type(marks))
'''if isinstance(marks, str):
    print ('Enter a valid number')'''
try:
    marks = float(marks)
    if float(marks) >= 90:
        print("Outstanding")
    elif float(marks) >= 75:
        print("Excellent")
    elif float(marks) >= 50:
        print("Pass")
    elif float(marks) < 50:
        print ("Fail")
except ValueError:
    print ('Enter a valid number')

```
