##Task-6: Create a pyspark dataframe df1 reading the data from the above file using pyspark magic command %python

```python
%python
from pyspark.sql.session import SparkSession
import pandas as pd
import numpy as np
import py
spark = SparkSession.builder.getOrCreate()
df = spark.read.csv("dbfs:/Volumes/workspace/default/mobile_metrics/mobile_os_usage.csv", header=True)
df.show()
```

##Task-7: Write the above dataframe df1 data into a databricks table called 'default.mobile_os_usage' using pyspark magic command %python

```python
%python
df = df.toDF(*[c.replace(" ", "_") for c in df.columns])
d1 = df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("mobile_os_usage")
df2 = spark.read.table("mobile_os_usage")
df2.display()
```

#Task-8: Write sql query to display the data loaded into the table 'default.mobile_os_usage' using the pyspark magic command %python

```python
%python
db_filtered = spark.sql("select * from mobile_os_usage")
db_filtered.display()
```
