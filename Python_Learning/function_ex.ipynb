```python
def bonus_calc (p_net_salary : float, p_rating : int):
    '''This function calculates bonus amount based on Salary Amount'''
    if p_rating == 5:
        bonus_amt = p_net_salary * 2
    elif p_rating == 4:
        bonus_amt = p_net_salary * 1.5
    elif p_rating ==  3:
        bonus_amt = p_net_salary * 1
    elif p_rating == 2:
        bonus_amt = p_net_salary * 0.5
    else:
        bonus_amt = 0
    return (bonus_amt/12)

def tax_calc (p_gross_salary : float):
    '''This function calculates tax amount based on Salary Amount'''
    if p_gross_salary <= 100000:
        tax_amt = 0
    elif p_gross_salary >= 100001 and p_gross_salary <= 125000:
        tax_amt = (125000-100001) * (10/100)
    elif p_gross_salary >= 125001 and p_gross_salary <= 150000:
        tax_amt = (150000-125001) * (20/100)
    else:
         tax_amt = (p_gross_salary-150001) * (30/100)
    return (tax_amt)

def net_salary (p_gross_salary : float):
    '''This function calculates net salary based on Salary Amount'''
    net_salary = p_gross_salary - tax_calc(p_gross_salary)
    return (net_salary)


gross_salary = float(input("Enter the Grooss Salary"))
rating = int(input("Enter the Rating 1 - 5"))
tax = tax_calc(gross_salary)
bonus = bonus_calc(net_salary(gross_salary), rating)
print (net_salary(gross_salary))
net = net_salary(gross_salary) + bonus
print (f'Gross Salary: {gross_salary:.2f}')
print (f'Rating: {rating}')
print (f'Tax: {tax:.2f}')
print (f'Bonus: {bonus:.2f}')
print (f'Net Salary: {net:.2f}')
```

** Use Case - 3 Number Utility Tool **

```python
def is_even(num : int):
    '''This function returns True if num is even'''
    if num % 2 == 0:
        return True
    else:
        return False

def perform_operation(num : int, operation : str):
    '''This function performs operation on num based on the given operation'''
    if operation.upper() == 'SQUARE':
        return (num ** 2)
    elif operation.upper() == 'CUBE':
        return (num ** 3)
    else:
        return('invalid operation')

def max_number (*args):  #Arbitrary arguments (Dynamic Arguments) - Positional
    '''This function returns the maximum number'''
    return(max(args))

print (is_even(10))
print (perform_operation(5, 'SQUARE'))
print (perform_operation(5, 'CUBE'))
print (perform_operation(5, 'UNKNOWN'))
print (max_number(10,20,50,100,90,70))
#print (max_number(a=10,b=20,c=50,d=100,e=90,f=70)) """get error while trying to send named/keyword parameter TypeError: max_number() got an unexpected keyword argument 'a'"""
```

** FBP (Function Based Program) calling below Methods **

1. Positional Arguments/Parameters
2. Named Arugments/Parameters
3. Default Value
4. Arbitrary Arguments(Dynamic Arguments)- Positional Arguments- (*args) Passing value as Tuple Collection type
5. Arbitrary Arguments(Dynamic Arguments)- Named/Keyword Arguments- (**kwargs) Passing value as Dictionary type

```python
# Example for Positional Arguments/Parameters & Named Arugments/Parameters
def subtract(a: int, b: int):
    '''This function subtract function'''
    return a-b

# calling as Positional Arguments
a=1; b=2;
print(a,b)
xy=subtract(a,b)
print (xy)
# calling as Named Arguments
a=1; b=2;
print(a,b)
xy=subtract(b,a)
print (xy)

# Example for Default Value
def addition(a: int, b: int = 1):
    return a+b

a=10; b=20;
xy=addition(a,b) #Passing both a and b.  passing b value overrides the default value
print(xy)
xy=addition(a)  #Passing a value without b argument.  default value used which was defined in function definition.
print (xy)

#Arbitrary Arguments(Dynamic Arguments)- Positional Arguments- (*args) Passing value as Tuple Collection type
def max_num (*args):
    '''This function returns the maximum number'''
    return(max(args))
print(max_num(10,20,50,100,90,70))

#Arbitrary Arguments(Dynamic Arguments)- Named/Keyword Arguments- (**kwargs) Passing value as Dictionary type
def max_exp_domain_level(**kwargs):
    print(type(kwargs))
    print(kwargs)
    for domain, exp in kwargs.items():
        print(f'(Maximum Years of Experience in {domain}: {max(exp)}')
max_exp_domain_level(java=[12,10,15], oracle=[11,20,16], bigdata=[2,4,6])
```
