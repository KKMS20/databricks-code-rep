** Use Case 1: Banking Eligibility Check **

```python
age = int(input("Enter your age: "))
income = int(input ("Enter your income: "))
if age < 18:
    print ("Not eligible for a bank account.")
elif age >=18 and income <15000:
    print ("Eligible for basic savings account.")
elif age >=18 and income >=15000 and income <= 50000:
    print ("Eligible for savings + salary account.")
elif age >=18 and income > 50000:
    print ("Eligible for premium account.")
```

** Portal Use Case-1: Transportation Selection **

```python
wallet_amount = 20000
bus_fare = 15000
train_fare = 17000
flight_fare = 19000
if wallet_amount >= bus_fare or wallet_amount >= train_fare or wallet_amount >= flight_fare:
    print ("Proceed to" , end = " ")
    if wallet_amount >= flight_fare:
        print ("Flight Travel")
    elif wallet_amount >= train_fare:
        print ("Train Travel")
    else:
        print ("Bus Travel")
else:
    print ("Can not Travel, Since the Wallet Amount is less than Bus Fare or Train Fare or Flight Fare")
```
