** Types of Operators **

1. Assignment
2. Compare/Conditional
3. Arithmetic
4. Logical

```python
a = 10 # '=' is Assignment Operator
if a==10: #True
    print ('a equal to 10 is True')

```

```python
# Compare/Conditional Operators
if a < 10: #True
    print ('True: a less than 10')
else: #False
    print ('False: a is not less than 10')

if a<=10: #True
    print ('True: a is less than or equal to 10')
else: #False
    print ('False: a is not less than or equal to 10')

if a > 10: #True
    print ('True: a greater than 10')
else: #False
    print ('False: a is not greater than 10')

if a>=10: #True
    print ('True: a is greater than or equal to 10')
else: #False
    print ('False: a is not greater than or equal to 10')

if a==10: #True
    print ('True: a is equal to 10')
else: #False
    print ('False: a is not equal to 10')

if a!=10: #True
    print ('True: a is not equal to 10')
else: #False
    print ('False: a is equal to 10')

```

**Python Operator: Use Case 1: Internet Data Usage Calculator**

```python
tot_mon_data_limit_in_gb = input ("Enter the Total Monthly Data Limit in GB")
data_used_so_for_in_gb = input ("Data used so for in GB")
remaining_data =int(tot_mon_data_limit_in_gb) - int(data_used_so_for_in_gb)
usage_percentage = (int(data_used_so_for_in_gb)/int(tot_mon_data_limit_in_gb))*100
print ('Total Monthly Data Limit in GB',tot_mon_data_limit_in_gb)
print ('Data used so for in GB',data_used_so_for_in_gb)
print ('Remaining Data in GB',remaining_data)
print ('Usage Percentage',f'{usage_percentage:.2f}')
if usage_percentage >=80:
    print ("Warning: High usage, consider upgrading your plan.")
```

** Python Operator: Use Case 2: Shopping Discount Calculation

```python
original_price = input("Enter Original Price")
discount_percentage = input("Enter Discount Percentage")
discount_amount = (float(original_price) * int(discount_percentage))/100
final_price = float(original_price) - discount_amount
#print (f"{original_price:.2f}")
print ('Original Price',f"{float(original_price):.2f}")
print('Discount Percentage',discount_percentage,'%')
print ('Discount Amount',f"{discount_amount:.2f}")
print ('Final Price',f"{final_price:.2f}")
```
