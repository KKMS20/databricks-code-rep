** Collections - List **

List is mutable (can be Modified).

List is Indexed and Sequenced.

List occupied more memory than Tuple.

Iteration over lists is time-consuming rather than Tuple.


```python
employee_name_list = ['Murugan', 'Kumaran','Karthick','Saravanan']
employee_age_list = [40, 50, 25, 35]
#prove that which is list data type
print (type(employee_name_list)); print (type(employee_age_list));
#prove that list type is indexed
print (employee_name_list[0]);print (employee_name_list[1]);
#prove that list is sequence/iterable
for name in employee_name_list:
    print (name)
for age in employee_age_list:
    print (age)
```

** Collection - Dictionary **

Collection of key-value pairs ({key: value}) not Indexed.

Used Curly Brackets.

Not Indexed but Sequenced.

Keys must be unique; values can be duplicates.



```python
employee_name_age_dict = {'Murugan' : 40, 'Kumaran' : 50, 'Karthick' : 25, 'Saravanan' : 35 }
#Prove that type is Dictionary
print (type(employee_name_age_dict))
#Prove that Dictionary is not Indexed and Key Based
#print (employee_name_age_dict[0])
print (employee_name_age_dict['Murugan'])
#Prove that Dictionary is sequence/iterable
print ('Printing Keys alone:')
for name in employee_name_age_dict.keys():
    print (name)
print ('Printing Values alone:')
for age in employee_name_age_dict.values():
    print (age)
print ('Printing Keys along with respective Values:')
for name, age in employee_name_age_dict.items():
    print (name, age)
```

** Collection - Tuple **

Tuple is immutable (Read Only- Can not be modified).

Tuple is also Indexed and Sequenced List.

Tuple occupied less memory than List.

Iteration over tuples is faster.

```python
#Indexed Sequenced collection of HETROGENEOUS (not mandatorily) elements/items
employee_details_tuple = ('Murugan', 40, 90000, 'Manager', 10)
#Prove that type is Tuple
print (type(employee_details_tuple))
#Prove that Tuple is  Indexed
print (employee_details_tuple[0]);print (employee_details_tuple[1]);
#Prove that Dictionary is sequence/iterable
for emp in employee_details_tuple:
    print(emp)
for emp in employee_details_tuple:
    print(str(emp)[0])
```

** Collection - Set **

#set {value1,value2,value3} - Sequenced collection of DE-DUPLICATED HOMOGENEOUS (not mandatorily) elements/items

#Set is used for performing DE-DUPLICATION & performing set operations (union, intersect, minus) (we will learn in level2 learning)

#Not indexed

```python
aspirants_set_sql1 = {'Kumaran', 'Saravanan', 'Karthick', 'Murugan'}
aspirants_set_sql2 = {'Kandan', 'Velan','Palani', 'Kumaran', 'Senthil', 'Velan'}
print (aspirants_set_sql1)
print (aspirants_set_sql2)
print (type(aspirants_set_sql1))
print (type(aspirants_set_sql2))
#How to access the elements/items of a set?
#######Can not by Index.  Can access by only Sequence
#Indexed? No, i can't access the elements/items using index
print("Let us see set is indexed? no")
#print(aspirants_set_sql1[0]) #This will fail
for set_aspirant1 in aspirants_set_sql1:
    print (set_aspirant1)
for set_aspirant2 in aspirants_set_sql2:
    print (set_aspirant2)

print ('SET-1', aspirants_set_sql1)
print ('SET-2', aspirants_set_sql2)

#####SET OPERATORS
#aspirants_set_sql1 UNION aspirants_set_sql2
print('SET1 UNION SET2', aspirants_set_sql1.union(aspirants_set_sql2))
#aspirants_set_sql1 INTERSECT aspirants_set_sql2
print('SET1 INTERSECT SET2', aspirants_set_sql1.intersection(aspirants_set_sql2))
#aspirants_set_sql1 DIFFERENCE aspirants_set_sql2
print('SET1 DIFFERENCE SET2', aspirants_set_sql1.difference(aspirants_set_sql2))
#aspirants_set_sql1 DIFFERENCE aspirants_set_sql1
print('SET2 DIFFERENCE SET2', aspirants_set_sql2.difference(aspirants_set_sql1))
```
