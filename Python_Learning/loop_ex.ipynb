** Use Case-2: Write a program to process a Salary increment in a loop until the salary reaches a specific target amount. **

```python
starting_salary = 40000
final_salary = 80000
while starting_salary <= 80000:
    increment_salary = (starting_salary * 10/100)
    #print (increment_salary)
    if starting_salary + increment_salary <= final_salary:
        starting_salary = starting_salary + increment_salary
        print (starting_salary)
    else:
        starting_salary = starting_salary + increment_salary
        print('Your salary reached the target salary.. So no increment ')
```

** Use Case-3: Write a program to apply a 10% tax for all states. Use the continue statement to skip the tax calculation specifically for Pondy and Goa. **

```python
state_list=['TamilNadu', 'Pondy', 'Karnataka', 'Goa', 'Telugana', 'MadhyaPradesh']
state_tax=10
income=50000
for list in state_list:
    if list == 'Pondy' or list == 'Goa':
        print (f'No State Tax for {list} on {income}') # No Tax
        continue
    print (list)
    tax_on_income = income * (state_tax/100)
    print (f'{state_tax}% for {list} on {income} = {tax_on_income}')
```

**Use Case-2 in Task Sheet: Sum of Even and Odd Numbers - Write a program that asks the user for a positive integer n.  Using a loop, calculate and print: **

```python
n = int(input("Enter the whole integer number, to sum of odd / even numbers from 1 to till whole integer number"))
odd = 0;
even = 0;
for i  in range (1,n+1):
    if (i%2) == 0:
        even=even + i
    else:
        odd=odd + i
print (f'Sum of odd numbers: {odd}')
print (f'Sum of even numbers: {even}')
```

** Use Case-3 in Task Sheet: Table Generator - Nested Looping  Write a program that takes a list of values/(number and generate a range) from the user and prints the multiplication table from 1 to 10 for that list of numbers. **

```python
table_list =[2,4,6]
for i in table_list:
    print (f'Table Number: {i}')
    for j in range(1,11):
        print (f'{j} * {i} = {j*i}')
```
