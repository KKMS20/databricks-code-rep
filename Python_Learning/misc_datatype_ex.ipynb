** #Bytes & Memory view **

Both are not used directly, at the time of sending data across network or storing inside storages, bytes are used. Memory view is used to see the memory address of where these bytecodes are stored

```python
empname='Irfan'
print(empname)
print(type(empname))
network_empname=empname.encode()
print(network_empname)
print (memoryview(network_empname))
print (type(network_empname))

empsal=100000
print(empsal)
print(bytes(empsal))
print(type(empsal))
```
