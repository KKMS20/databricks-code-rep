** Standard Input and Output Functions **


```python
import pdb

from learnpythonpkg.fundapkg import pandas_ex

# Standard Input  -  The input function always returns String type.
aspirant_name = input("Enter Aspirant Name: ")
print(aspirant_name)
print(type(aspirant_name))
aspirant_number = input("Enter Aspirant Number: ")
print(aspirant_number)
print(type(aspirant_number))
aspirant_age = (input("Enter Aspirant Age: "))  #casting from str to int
print(aspirant_age)
if aspirant_age.isdigit() and int(aspirant_age)!=0:
    aspirant_age = int(aspirant_age)
    print(type(int(aspirant_age)))
else:
    print(type(aspirant_age))
```

```python
# Standard Output - This output function returns NoneType.
employee_name = 'Kumaran M'
employee_salary = 199000.89999

print('Employee Name', employee_name)  # Default Separator is space;
print(type(employee_name))

# Default Separator is space; in this print statement Separator changed from default value space to " : ".
# Default End Terminator is /n (new line); in this print statement End Terminator changed from default value "/n" (New Line) to " | "
print('Employee Name', employee_name, sep=" : ", end = " | ")
print(type(employee_name))
print(f"Employee Salary: {employee_salary:.2f}")
```
